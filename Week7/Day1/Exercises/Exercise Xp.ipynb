# Exercise 1: Introduction to Data Analysis

Data analysis is the process of collecting, cleaning, organizing, and examining data to find useful patterns and draw conclusions. It can involve simple calculations and charts or advanced techniques such as statistical modeling and machine learning. The goal is to turn raw information into evidence that helps people answer questions and make decisions.

Data analysis is important because modern organizations generate large amounts of information—from online purchases and customer feedback to medical records and sensor readings. Analyzing that information can reveal what is working, identify problems earlier, and help predict future needs. Current trends include using artificial intelligence and machine learning to find patterns, analyzing data in real time, and paying greater attention to data privacy and quality. These tools are useful only when the underlying data is reliable and handled responsibly.

Three areas where data analysis is widely applied are:

1. **Healthcare:** Hospitals and researchers analyze patient records and test results to identify health risks, improve diagnoses, and evaluate treatments. For example, analysis of disease trends can help public-health teams plan where to focus resources.
2. **Business and retail:** Companies study sales, customer preferences, and website activity to understand demand, improve products, and manage inventory. A retailer might use past sales and seasonal patterns to decide how much stock to order.
3. **Transportation:** Transit agencies and navigation services analyze traffic, travel times, and vehicle-location data to improve routes and reduce delays. Real-time traffic analysis can help drivers choose a faster route.

In short, data analysis helps people make better-informed decisions across many aspects of modern life. Its value depends not just on advanced technology, but also on asking clear questions, using trustworthy data, and considering privacy and fairness.

```python
def analyze_dataframe(df, name):
    print(f"=== Analyis of {name} ===")
    for col in df.columns:
        dtype = df[col].dtype
        # Simple rule: numeric types are quantitative, others (object, category) are qualitative
        if pd.api.types.is_numeric_dtype(dtype):
            category_type = "Quantitative (Numerical)"
            reasoning = f"Contains numerical values of type '{dtype}'."
        else:
            category_type = "Qualitative (Categorical)"
            reasoning = f"Contains textual, categorical, or mixed descriptive values of type '{dtype}'."
        print(f"- **{col}**: {category_type} | Reason: {reasoning}")
    print("\n")

analyze_dataframe(data, "data (clean_dataset.csv)")
analyze_dataframe(data2, "data2 (Mental health Depression disorder Data)")
analyze_dataframe(data3, "data3 (Time Americans Spend Sleeping)")
```

### **Exercise 6: Identifying Data Types**

* **A company’s financial reports stored in an Excel file.**
  * **Type**: **Structured Data**
  * **Reasoning**: The data is neatly organized into defined rows, columns, and tables, which makes it easily searchable and predictable.

* **Photographs uploaded to a social media platform.**
  * **Type**: **Unstructured Data**
  * **Reasoning**: Photographs consist of raw pixel data with no predefined data model, making them unstructured even if they contain some structured metadata.

* **A collection of news articles on a website.**
  * **Type**: **Unstructured Data**
  * **Reasoning**: News articles are written in natural language, which is free-form text that does not fit into a strict tabular format.

* **Inventory data in a relational database.**
  * **Type**: **Structured Data**
  * **Reasoning**: Relational databases enforce a highly organized schema with rigid tables, relationships, and data types.

* **Recorded interviews from a market research study.**
  * **Type**: **Unstructured Data**
  * **Reasoning**: Audio files contain wave signals of speech and lack any structured data format until they are transcribed, labeled, or parsed.

### **Exercise 7: Transformation Exercise**

For each unstructured data source below, here are the proposed methods to convert them into structured data along with the reasoning:

#### **1. A series of blog posts about travel experiences.**
* **Transformation Method**:
  1. Use **Named Entity Recognition (NER)** to extract specific entities such as destinations (City, Country), landmarks visited, travel dates, and transport types.
  2. Apply **Sentiment Analysis** to rate the traveler's experience on a scale (e.g., -1 for negative, 0 for neutral, 1 for positive).
  3. Use **Topic Modeling** (such as LDA or LLMs) to categorize posts into tags (e.g., 'Adventure', 'Budget', 'Luxury', 'Food').
* **Resulting Schema**: `[Post_ID, Author, Destination_City, Destination_Country, Sentiment_Score, Category_Tags]`
* **Reasoning**: Travel blogs contain valuable demographic, geographic, and emotional insights. Standardizing text into discrete columns allows businesses to filter, rank, and analyze popular destinations quantitatively.

---

#### **2. Audio recordings of customer service calls.**
* **Transformation Method**:
  1. Pass the raw audio files through an **Automatic Speech Recognition (ASR) / Speech-to-Text** pipeline to generate textual transcripts.
  2. Parse the metadata of the call (timestamp, call duration, agent ID, customer caller ID).
  3. Apply **Natural Language Processing (NLP)** on the transcripts to extract:
     * *Sentiment polarity* (beginning of call vs. end of call).
     * *Primary issue/category* (e.g., 'Billing', 'Technical Support', 'Cancellation').
     * *Resolution status* (e.g., 'Resolved', 'Escalated').
* **Resulting Schema**: `[Call_ID, Timestamp, Agent_ID, Customer_ID, Call_Duration_Sec, Issue_Category, Initial_Sentiment, End_Sentiment, Is_Resolved]`
* **Reasoning**: Converting vocal patterns and dialogs into a structured record makes it possible to aggregate metrics like Average Handle Time (AHT) or Customer Satisfaction (CSAT) trends over time.

---

#### **3. Handwritten notes from a brainstorming session.**
* **Transformation Method**:
  1. Scan or photograph the notes and process them using **Intelligent Character Recognition (ICR) / Optical Character Recognition (OCR)** software specialized in handwriting.
  2. Use a LLM-based parsing step to organize raw text into structured lists (e.g., grouping bullet points under corresponding main headings/topics).
  3. Map the extracted action items to specific owners and set deadlines.
* **Resulting Schema**: `[Idea_ID, Category/Theme, Proposed_Concept, Contributor, Action_Item, Assigned_Owner, Deadline]`
* **Reasoning**: Brainstorming notes are useful only when acted upon. Transforming them into a structured task tracker allows teams to query, prioritize, and assign ownership programmatically.

---

#### **4. A video tutorial on cooking.**
* **Transformation Method**:
  1. Extract the audio track and use **Speech-to-Text** to transcribe spoken instructions.
  2. Use **Computer Vision (Object Detection & Action Recognition)** to detect ingredients shown on screen and associate time stamps with specific actions (e.g., 'chopping', 'boiling').
  3. Structure the output into a sequential dataset of steps, ingredients, measurements, and cooking durations.
* **Resulting Schema**: `[Recipe_ID, Recipe_Name, Prep_Time_Min, Cook_Time_Min, Ingredients_List, [Step_Number, Action, Duration_Min]]`
* **Reasoning**: Raw video is difficult to search or quickly read. Converting video frames and voiceovers into a structured recipe card format allows users to filter recipes by ingredients, total cooking time, or skill level easily.
