```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# Helper to print distinct headers for exercises
def print_header(title):
    print("\n" + "="*80)
    print(f" {title} ".center(80, "="))
    print("="*80 + "\n")

# =========================================================================
# Exercise 1: Basic Data Analysis with Kaggle (California Housing)
# =========================================================================
print_header("Exercise 1: California Housing Basic Data Analysis")
# Let's use the california housing train dataset located at /content/sample_data/
cal_df = pd.read_csv('/content/sample_data/california_housing_train.csv')
print("Loaded California Housing Dataset. Shape:", cal_df.shape)

target_col = 'median_house_value'
mean_val = cal_df[target_col].mean()
median_val = cal_df[target_col].median()
std_val = cal_df[target_col].std()

print(f"Target Column: '{target_col}'")
print(f"• Mean: {mean_val:,.2f}")
print(f"• Median: {median_val:,.2f}")
print(f"• Standard Deviation: {std_val:,.2f}")

# =========================================================================
# Exercise 2: Comparative Analysis of Retail Data
# =========================================================================
print_header("Exercise 2: Comparative Analysis of Retail Data & Unstructured Reviews")

# 2a. Analyze Retail Dataset
retail_sales = pd.read_csv("/content/Retail Data Analytics dataset/sales data-set.csv")
retail_features = pd.read_csv("/content/Retail Data Analytics dataset/Features data set.csv")
retail_stores = pd.read_csv("/content/Retail Data Analytics dataset/stores data-set.csv")

# Convert date
retail_sales['Date'] = pd.to_datetime(retail_sales['Date'], dayfirst=True)
print("Retail Sales (Head):")
print(retail_sales.head(3))
print("\nWeekly Sales Trend Summary (Aggregated by Store Type):")
merged_retail = retail_sales.merge(retail_stores, on='Store', how='inner')
print(merged_retail.groupby('Type')['Weekly_Sales'].describe())

# 2b. Analyze Clothing Reviews
reviews_df = pd.read_csv("/content/Women's E-Commerce Clothing Reviews/Womens Clothing E-Commerce Reviews.csv")
print("\nWomen's E-Commerce Clothing Reviews (Head):")
print(reviews_df[['Review Text', 'Rating', 'Recommended IND']].head(3))

# Basic sentiment extraction from Rating
avg_rating = reviews_df['Rating'].mean()
rec_rate = reviews_df['Recommended IND'].mean() * 100
print(f"\n• Average Rating: {avg_rating:.2f}/5")
print(f"• Recommendation Rate: {rec_rate:.2f}%")

# Extract most frequent words in positive reviews as simple text topic mining
positive_reviews = reviews_df[reviews_df['Rating'] >= 4]['Review Text'].dropna().str.lower()
words = " ".join(positive_reviews).split()
stopwords = {'the', 'and', 'a', 'to', 'it', 'i', 'is', 'of', 'in', 'for', 'this', 'with', 'but', 'on', 'my', 'that', 'so', 'was'}
filtered_words = [w for w in words if w.isalpha() and w not in stopwords]
word_counts = pd.Series(filtered_words).value_counts().head(5)
print("\nTop topics/frequently mentioned words in Positive Reviews:")
print(word_counts)

print("\nComparative Analysis / Unstructured Data Challenges:")
print("Challenges: Natural language reviews are unstructured, containing inconsistent grammar, spellings, and context.")
print("Comparison: While the Retail dataset yields precise numeric aggregates (like sales metrics), the unstructured clothing reviews reveal qualitative dimensions (e.g., product fit issues or styling details) that raw numbers cannot convey.")

# =========================================================================
# Exercise 3: Basic Data Exploration in E-Commerce
# =========================================================================
print_header("Exercise 3: E-Commerce Basic Data Exploration")

ecom_df = pd.read_csv("/content/E-Commerce Dataset/data.csv", encoding='ISO-8859-1')
print("E-Commerce Data Head:")
print(ecom_df.head(3))
print("\nDataset Info:")
ecom_df.info()

print("\nStructured Data Columns:")
print("• InvoiceNo, StockCode, Quantity, InvoiceDate, UnitPrice, CustomerID, Country")
print("\nUnstructured Data Complement Suggestion:")
print("• Customer textual product reviews and detailed search queries can complement this dataset.")
print("• Integration: Sentiment scores extracted from customer reviews could be mapped against customer return rates or product discount thresholds to optimize product recommendations.")

# =========================================================================
# Exercise 4: Analyzing a Public Transportation Dataset with Focus on Data Types
# =========================================================================
print_header("Exercise 4: Metro Interstate Traffic Volume Data Analysis")

traffic_df = pd.read_csv("/content/Metro Interstate Traffic Volume/Metro_Interstate_Traffic_Volume.csv")
print("Metro Traffic Head:")
print(traffic_df.head(3))

print("\nCategorization of Elements:")
print("1. Date-Time ('date_time'): Structured (Highly standard formatted timestamps, easily parseable for time series analysis)")
print("2. Traffic Volume ('traffic_volume'): Structured (Numerical continuous data representing measurable vehicles)")
print("3. Weather Description ('weather_description'): Semi-Structured / Categorical (Phrases like 'scattered clouds' represent fixed descriptors, but are expressed in raw strings)")

# =========================================================================
# Exercise 5: Basic Data Analysis in a Movie Ratings Dataset
# =========================================================================
print_header("Exercise 5: MovieLens Ratings Data Analysis")

ratings_df = pd.read_csv("/content/MovieLens 20M Dataset/rating.csv")
print("MovieLens Ratings Head:")
print(ratings_df.head(3))

print("\nStructured elements: userId, movieId, rating, timestamp.")
print("Why they are structured: They conform to a strict tabular schema with explicit formats (integers, floats, unix timestamps) requiring no raw parsing or semantic natural language interpretation.")

# =========================================================================
# Exercise 6: Creating a Synthetic Product Catalog
# =========================================================================
print_header("Exercise 6: Generating Synthetic Product Catalog")

# Install Faker if not present
try:
    from faker import Faker
except ImportError:
    !pip install Faker
    from faker import Faker

fake = Faker()
np.random.seed(42)
Faker.seed(42)

products_list = []
for i in range(1, 501):
    product = {
        "Product_ID": f"PROD_{i:04d}",
        "Name": fake.catch_phrase(),
        "Description": fake.paragraph(nb_sentences=3),
        "Price": round(np.random.uniform(5.99, 499.99), 2)
    }
    products_list.append(product)

catalog_df = pd.DataFrame(products_list)
print("Generated Synthetic Catalog successfully! Shape:", catalog_df.shape)
print("\nCatalog Sample:")
print(catalog_df.head(5))
```
